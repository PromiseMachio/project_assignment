## Data cleaning


```python
## Removing  columns with many misssing values 
# check missing values per column
missing = df.isnull().sum().sort_values(ascending=False)
print(missing.head(20))

# drop columns with > 70% missing values
df = df.dropna(axis=1, thresh=len(df)*0.3)
```

## Data preparation

```python
# changing the date from object to a valid 
df['publish_time'] = pd.to_datetime(df['publish_time'], errors='coerce')
df['year'] = df['publish_time'].dt.year
```

```python
# Creating new columns 
df['abstract_word_count'] = df['abstract'].fillna("").apply(lambda x: len(x.split()))
```

```python
# final clean up
df = df.drop_duplicates()
# Reseting indexs
df = df.reset_index(drop=True)

```

## Data Visualization

```python
%matplotlib inline
import matplotlib.pyplot as plt


year_counts = df['year'].value_counts().sort_index()
print(year_counts)



plt.figure(figsize=(8,5))
plt.bar(year_counts.index, year_counts.values)
plt.title("Publications by Year")
plt.xlabel("Year")
plt.ylabel("Number of Papers")
plt.show()
```

```python
#Top journals 
top_journals = df['journal'].value_counts().head(10)
print(top_journals)

plt.figure(figsize=(10,6))
top_journals.plot(kind='bar')
plt.title("Top Journals Publishing COVID-19 Research")
plt.xlabel("Journal")
plt.ylabel("Number of Papers")
plt.xticks(rotation=45)
plt.show()
```

```python
#Visualization by bar word count
from collections import Counter
import re

# join all titles into one string
all_titles = " ".join(df['title'].dropna().astype(str))

# simple cleanup: lowercase and remove non-letters
words = re.findall(r'\b[a-z]{3,}\b', all_titles.lower())

# count words
word_freq = Counter(words).most_common(20)
print(word_freq)

# plot top words
words, counts = zip(*word_freq)
plt.figure(figsize=(10,5))
plt.bar(words, counts)
plt.title("Most Frequent Words in Paper Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
# Visualization by wordcloud
from wordcloud import WordCloud

wordcloud = WordCloud(width=800, height=400, background_color='white').generate(all_titles)

plt.figure(figsize=(10,6))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```

```python
# Visualization of distribution of paper by source
source_counts = df['source_x'].value_counts().head(10)

plt.figure(figsize=(8,5))
source_counts.plot(kind='bar')
plt.title("Distribution of Papers by Source")
plt.xlabel("Source")
plt.ylabel("Count")
plt.show()
```
